```python
def multiply(sys, sys_degree, M_degree, nb_of_variables):
    """
    input: 
    sys: a list of equations, 
    sys_degree: an int representing the degree of the given system
    M_degree: an int representing the degree of the generated macauley matrix
    nb_of_variables: the number of variables present in the given system
    Output: 
    a full or partial Macauley matrix of degree M_degree.
    """
    x = PolynomialRing(ZZ, 2, 'x').gens()
    res = [eq for eq in sys]
    for eq in sys:
        for dd in range(sys_degree, M_degree):
            for pair in generate_monomials_of_degree(M_degree-dd, nb_of_variables, x):
                new_eq = eq*pair[0]*pair[1]
                res.append(new_eq)
    return res
```
