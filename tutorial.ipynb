```python
# select a random element
q = 9
F = GF(q)
print("random element from GF(9):", F.random_element())


# gens()
"""
Many parent objects in Sage are equipped with generators, 
which are special elements of the object. 
For example, the polynomial ring Z[x, y, z] is generated by x, y, and z. 
In Sage the -th generator of an object X is obtained using the notation X.gen(i). 
From the Sage interactive prompt, the shorthand notation X.gen(i) is also allowed.
"""
n = 8
R = PolynomialRing(ZZ, n, 'x')
print(f"generators for PolynomialRing(ZZ, {n}, 'x')", R.gens())
print('The one-but-last generator is accessed as follows:', R.gens()[n-1])



# vector space
q = 256
dim = 3
V = VectorSpace(GF(q), dim)
print(f'A basis for VectorSpace(GF({q}), {dim}', V.basis())
print(f'The generators of the above declared vector space are:',V.gens())
```

```python
# get the coefficients of the variables with specified degrees in a polynomial
P.<x, y, z> = PolynomialRing(ZZ, 3)
length = P.gens()
print(P.gens())
degrees = {P.gens()[i]: 1 for i in range(len(length))} 
print(degrees)
f = x^2 + y + z
print(f.coefficients())
print(f.coefficient(degrees))
print(f.coefficient({x: 0, y: 1, z: 0}))
```
